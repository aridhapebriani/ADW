## **2. Plot ACF Deret MA(1)**

Membuat plot ACF dari masing-masing deret MA(1) yang diperoleh pada Nomor 1 untuk membandingkan pola autokorelasi, khususnya pada lag 1, berdasarkan perubahan nilai parameter θ.

### a. Plot ACF MA(1) dengan θ = 0.1

```python
plot_acf(ma_01, lags=10, zero=False)
plt.title("ACF MA(1), theta = 0.1")
plt.show()
```

### b. Plot ACF MA(1) dengan θ = 0.3

```python
plot_acf(ma_03, lags=10, zero=False)
plt.title("ACF MA(1), theta = 0.3")
plt.show()
```

### c. Plot ACF MA(1) dengan θ = 0.7

```python
plot_acf(ma_10, lags=10, zero=False)
plt.title("ACF MA(1), theta = 0.7")
plt.show()
```

### d. Plot ACF MA(1) dengan θ = 1.0

```python
plot_acf(ma_10, lags=10, zero=False)
plt.title("ACF MA(1), theta = 1.0")
plt.show()
```

### e. Plot ACF MA(1) dengan θ = 1.5

```python
plot_acf(ma_15, lags=10, zero=False)
plt.title("ACF MA(1), theta = 1.5")
plt.show()
```

### f. Plot ACF MA(1) dengan θ = 2.0

```python
plot_acf(ma_20, lags=10, zero=False)
plt.title("ACF MA(1), theta = 2.0")
plt.show()
```

Berdasarkan plot ACF, nilai autokorelasi pada lag 1 cenderung semakin negatif ketika parameter θ meningkat dari 0,1 hingga 1,0. Pada θ = 1,0, nilai autokorelasi lag 1 mencapai nilai negatif terbesar. Setelah θ lebih besar dari 1, nilai absolut autokorelasi lag 1 kembali menurun. Pola ini sesuai dengan autokorelasi teoritis MA(1), yaitu

$$ \rho_1=\frac{-\theta}{1+\theta^2}. $$

Perbedaan antara nilai autokorelasi sampel pada plot dan nilai teoritis dapat terjadi karena panjang pengamatan yang digunakan hanya 20.

# **5. Simulasi Deret MA(2)**

5. Buatlah simulasi deret MA (2) dengan panjang pengamatan 50 dengan parameter:
   
   a. $\theta_1=0.3, \theta_2=0.8$
   
   b. $\theta_1=-0.3, \theta_2=0.8$
   
   c. $\theta_1=0.3, \theta_2=-0.8$
   
   d. $\theta_1=-0.3, \theta_2=-0.8$


```python
# a. θ1 = 0.3, θ2 = 0.8
JMLH_NIM = 104
np.random.seed(JMLH_NIM)
ar1 = np.array([1])

ma1 = np.array([1, -0.3, -0.8])
ma_object_a = ArmaProcess(ar1, ma1)
simulasi_a = ma_object_a.generate_sample(nsample=50)


# b. θ1 = -0.3, θ2 = 0.8
np.random.seed(JMLH_NIM)

ma1 = np.array([1, 0.3, -0.8])
ma_object_b = ArmaProcess(ar1, ma1)
simulasi_b = ma_object_b.generate_sample(nsample=50)


# c. θ1 = 0.3, θ2 = -0.8
np.random.seed(JMLH_NIM)

ma1 = np.array([1, -0.3, 0.8])
ma_object_c = ArmaProcess(ar1, ma1)
simulasi_c = ma_object_c.generate_sample(nsample=50)


# d. θ1 = -0.3, θ2 = -0.8
np.random.seed(JMLH_NIM)

ma1 = np.array([1, 0.3, 0.8])
ma_object_d = ArmaProcess(ar1, ma1)
simulasi_d = ma_object_d.generate_sample(nsample=50)
```

# **10. Simulasi Deret AR(2)**

Buatlah simulasi deret AR (2) dengan panjang pengamatan 100 dengan parameter:



  a. $\phi_1=1.5, \phi_2=-0.5$

```python
np.random.seed(104)

psi1_a, psi2_a = 1.5, -0.5

ar2_a = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(100):
    ar2_a.append(
        psi1_a * ar2_a[i+1] +
        psi2_a * ar2_a[i] +
        np.random.normal(0,1)
    )
ar2_a
```

```python
plt.figure(figsize=(10,5))
plt.plot(ar2_a, '-o')
plt.xlabel('Waktu')
plt.ylabel(r'$Y_t$')
plt.title(r'Simulasi AR(2): $\phi_1=1.5,\ \phi_2=-0.5$')
plt.show()
```

  b. $\phi_1=0.75, \phi_2=-0.25$

```python
np.random.seed(104)

psi1_b, psi2_b = 0.75, -0.25

ar2_b = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(100):
    ar2_b.append(
        psi1_b * ar2_b[i+1] +
        psi2_b * ar2_b[i] +
        np.random.normal(0,1)
    )
ar2_b
```

  c. $\phi_1=1, \phi_2=-0.25$

```python
np.random.seed(104)

psi1_c, psi2_c = 1.0, -0.25

ar2_c = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(100):
    ar2_c.append(
        psi1_c * ar2_c[i+1] +
        psi2_c * ar2_c[i] +
        np.random.normal(0,1)
    )
ar2_c
```
